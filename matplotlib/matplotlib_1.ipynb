### 2D Line plot

- Bivariate Analysis

```python
# plotting a simple functions
price = [20000, 34000, 45000, 40000, 69000, 65000]
year = [2015, 2016, 2017, 2018, 2019, 2020]
plt.plot(year, price) # (year -> for x axis, price -> for y axis)
```

```python
# from a pandas dataframe
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])
```

```python
# label title
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('rohit sharma vs virat kohli runs', color='black')
plt.xlabel('seasons')
plt.ylabel('runs scored')
```

```python
# color(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['V Kohli'], color = '#F54927')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#2731F5')

plt.title('rohit sharma vs virat kohli runs', color='black')
plt.xlabel('seasons')
plt.ylabel('runs scored')
```

```python
# color(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'], batsman['V Kohli'], color = '#F54927', linestyle='dashdot', linewidth=3)
plt.plot(batsman['index'], batsman['RG Sharma'], color='#2731F5', linestyle='dashed', linewidth=2)

plt.title('rohit sharma vs virat kohli runs', color='black')
plt.xlabel('seasons')
plt.ylabel('runs scored')
```

```python
# marker(size)
plt.plot(batsman['index'], batsman['V Kohli'], color = '#F54927', linestyle='dashdot', linewidth=3, marker='D', markersize=5)
plt.plot(batsman['index'], batsman['RG Sharma'], color='#2731F5', linestyle='dashed', linewidth=2, marker='o', markersize=8)

plt.title('rohit sharma vs virat kohli runs', color='black')
plt.xlabel('seasons')
plt.ylabel('runs scored')
```

```python
# legend -> location
plt.plot(batsman['index'], batsman['V Kohli'], color = '#F54927', linestyle='dashdot', linewidth=3, marker='D', markersize=5, label = 'virat')
plt.plot(batsman['index'], batsman['RG Sharma'], color='#2731F5', linestyle='dashed', linewidth=2, marker='o', markersize=8, label ='rohit')

plt.title('rohit sharma vs virat kohli runs', color='black')
plt.xlabel('seasons')
plt.ylabel('runs scored')
plt.legend() # hidden parameter for location 'loc' initial loc='best'
```

```python
# plotting a simple functions
price = [20000, 34000, 45000, 40000, 69000, 65000, 650000]
year = [2015, 2016, 2017, 2018, 2019, 2020, 2021]
plt.plot(year, price) # (year -> for x axis, price -> for y axis)

plt.ylim(0,100000)
plt.xlim(2017, 2019)
```

### Bar chart

image.png

```python
# simple bar chart
children = [10,20,30,40,50]
color = ['red', 'green', 'blue','yellow', 'pink']
plt.bar(color, children, color=['red','green', 'blue','yellow', 'pink'])
plt.title('fav color')
plt.xlabel('colors')
plt.ylabel('no of student')
```

```python
# horizontal bar chart
plt.barh(color, children, color=['red','green', 'blue','yellow', 'pink'])
plt.title('fav color')
plt.xlabel('colors')
plt.ylabel('no of student')
```

```python
# multiple bar charts
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='green')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
# problem & solutions
children = [10,20,30,40,50]
color = ['red red red red red', 'green green green green green', 'blue blue blue blue blue','yellow yellow yellow yellow yellow', 'pink pink pink pink pink']
plt.bar(color, children, color=['red','green', 'blue','yellow', 'pink'])
plt.xticks(rotation='vertical')
plt.title('fav color')
plt.xlabel('colors')
plt.ylabel('no of student')
```

```python
# stacked bar chart
plt.bar(df['batsman'], df['2015'], label='2015', color='red')
plt.bar(df['batsman'], df['2016'], bottom=df['2015'], label='2016', color='green')
plt.bar(df['batsman'], df['2017'], bottom=df['2015'] + df['2016'], label='2017', color='blue')

plt.legend()
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.title('Runs by Year (Stacked)')
plt.show()
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
```

```python
# dataset
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%')
plt.show()
```

```python
# explode & shadow
plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode=[0.1,0,0.2,0,0.2,0], shadow=True)
plt.show()
```
