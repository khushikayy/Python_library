### What is Pandas
Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language.

https://pandas.pydata.org/about/index.html

### Panda Series
A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

#### Importing Pandas

```python
import numpy as np
import pandas as pd
```

#### Series from Lists

```python
# string
country = ['india', 'usa', 'china', 'dubai']
pd.Series(country)
```

```python
# integer
runs = [12, 67,69,90,45,100]
runs_series = pd.Series(runs)
```

```python
# custom index
marks = [76,56,89,100]
subject = ['math','english', 'hindi', 'science']
pd.Series(marks, index=subject)
```

```python
# setting a name
pd.Series(marks, index=subject, name="rishav's marks")
```

#### Series from dict

```python
marks = {
    "math" : 76,
    "english" : 56,
    "hindi" : 89,
    "science" : 100
}
marks_series = pd.Series(marks, name='marks')
marks_series
```
