### Selecting rows from a DataFrame


*   iloc - searches using index positions
*   loc - searches using index label


```python
# make a new dataframe from dist
students_dict = {
    'name': ['nitish','ankit','rupesh', 'rishabh','amit', 'ankita'],
    'iq': [100,90,120,80,0,0],
    'marks': [80,70,100,50,0,0],
    'package' : [10,7,14,2,0,0]

}

students = pd.DataFrame(students_dict)
students.set_index('name', inplace = True)
students
```

```python
# single row
movies.iloc[0]
```

```python
# multiple rows
movies.iloc[0:5]
```

```python
# fancy indexing
movies.iloc[[0,3,6,8,11]]
```

```python
# loc
students.loc['nitish']
```

```python
students.loc['nitish':'rishabh']
```

```python
students
```

```python
students.iloc[0]
```
