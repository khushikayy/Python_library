```python
# agg method -> passing dict
# genres.sum(numeric_only = True)
genres.agg(
    {
        'Runtime':'mean',
        'IMDB_Rating':'mean',
        'No_of_Votes': 'sum',
        'Gross' : 'sum',
        'Metascore': 'mean'
    }
)
```

```python
# passing list
genres[['Runtime', 'IMDB_Rating', 'No_of_Votes', 'Gross', 'Metascore']].agg(
    ['min', 'max', 'mean', 'sum']
)
```

```python
genres.agg(
    {
        'Runtime':['min','mean'],
        'IMDB_Rating':'mean',
        'No_of_Votes': ['sum', 'max'],
        'Gross' : ['sum', 'max'],
        'Metascore': 'mean'
    }
)
```

```python
# looping on groups
for group, data in genres:
  # print(type(group),type(data))
  print(data)
```

```python
# looping on groups
rows = []

for group, data in genres:
    rows.append(data[data['IMDB_Rating'] == data['IMDB_Rating'].max()])

df = pd.concat(rows, ignore_index=True)
df
```

```python
# split (apply) combine
# apply -> built-in function
genres.apply(min)
```

```python
# find number of movies starting with A for each group
def foo(group):
  return group['Series_Title'].str.startswith('A').sum()
```

```python
genres.apply(foo)
```

```python
# find ranking of each movies in the group according to IMDB score
def rank_movies(group):
  group['genres_rank'] = group['IMDB_Rating'].rank(ascending=False)
  return group
```

```python
genres.apply(rank_movies)
```

```python
# find normalized IMDB rating group wise
def normalized(group):
  group['normalized_ranking'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min()) / (group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group
```

```python
genres.apply(normalized)
```

```python
for group, data in genres:
    print("Genre:", group)
    display(data)
```

### practice questions

```python
ipl = pd.read_csv('/content/deliveries.csv')
```

```python
ipl.head()
```

```python
ipl.shape
```

```python
# find the top 10 batsman in terms of runs
batsman = ipl.groupby('batsman')
```

```python
batsman['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
six = ipl[ipl['batsman_runs'] == 6]
```

```python
six.groupby('batsman')['total_runs'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
temp_df = ipl[ipl['over'] > 15]
```

```python
temp_runs = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
```

```python
temp_runs.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
# find V kohli's record against all team
batsman = ipl[ipl['batsman'] == 'V Kohli']
```

```python
batsman.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(1)
```

```python
# create a functions that can return the highest score of any batsman
def high_score(batsman):
  temp_df = ipl[ipl['batsman'] == batsman]
  return temp_df.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]
```

```python
high_score('V Kohli')
```

```python
high_score('DA Warner')
```

```python
high_score('MS Dhoni')
```
