```python
# problem in vectorized operation in vanilla python
s = ['cat', 'mat', None, 'rat']
# [i.startswith('c') for i in s] # this will raise a error for type none bcz the none type has no attribute startswith
```

```python
# how pandas solve this error
s = pd.Series(['cat', 'mat', None, 'rat'])
# string accesssor -> str
s.str.startswith('c')
# fast and optimized
```

```python
df = pd.read_csv('/content/titanic.csv')
df.head()
```

```python
df['Name']
```

```python
# Common Functions
# lower/upper/capitalize/title
df['Name'].str.lower()
df['Name'].str.upper()
df['Name'].str.capitalize()
df['Name'].str.title()
# len
df['Name'][df['Name'].str.len() == 82].values[0]
# strip
"                        rishav             ".strip()
df['Name'].str.strip()
```

```python
# split -> get
df['lastname'] = df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title', 'firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n = 1, expand = True)
```

```python
df.head()
df['title'].value_counts()
```

```python
# replace
df['title'] = df['title'].str.replace('Ms.', 'Miss.')
df['title'] = df['title'].str.replace('Mlle.', 'Miss.')
```

```python
df['title'].value_counts()
```

```python
# filering
# startswith/endswith
df[df['firstname'].str.startswith('A')]
df[df['firstname'].str.endswith('A')]
```

```python
# apply regex
# contains
# search john -> both case
df[df['firstname'].str.contains('john', case = False)].shape
# find lastname with start and end char vowel
df[df['lastname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```
