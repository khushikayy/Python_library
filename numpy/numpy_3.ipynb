### np.sort
Return a sorted copy of an array

```python
import numpy as np
a = np.random.randint(1,100,15)
print(a)
```

```python
print(np.sort(a))
```

```python
print(np.sort(a)[::-1])
```

```python
b = np.random.randint(1,100,24).reshape(6,4)
print(b)
```

```python
print(np.sort(b))
```

```python
print(np.sort(b,axis=0))
```

```python
np.append(b,np.ones((b.shape[0],1)), axis=1)
```

### np.concatenate
np.concatenate() function concatenate a sequesnce of array along an existing array

```python
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)
print(c)
print(d)
```

```python
np.concatenate((c,d),axis=0) # 0 is for rows
```

```python
np.concatenate((c,d),axis=1) # 1 is for column
```

### np.where
The np.where() function return the indices of the element is an input array where the given condition is satisfied

```python
a = np.random.randint(1,100,15)
a
```

```python
# where the element is greater than 50
np.where(a > 50)
```

```python
# replace all value > 50 with 0
# np.where(condition, true, false)
np.where(a > 50 , 0, a)
```

```python
np.where(a % 2 == 0, 0, 1)
```

### np.corrcoef
